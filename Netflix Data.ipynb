```python
plt.figure(figsize=(15, 10))
yearly_content_distribution = df.groupby(['release_year', 'type']).size().unstack(fill_value=0)
for index, content_type in enumerate(yearly_content_distribution.columns):
    plt.subplot(3, 3, index + 2)
    plt.bar(yearly_content_distribution.index, yearly_content_distribution[content_type], color='#4F6F52')
    plt.title(f'{content_type} Releases over the Years')
    plt.xlabel('Release Year')
    plt.ylabel('Number of Releases')
    plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
from collections import Counter

genre_data = df['listed_in'].apply(lambda x: x.split(', '))

genre_frequency = Counter(genre for sublist in genre_data for genre in sublist)


top_ten_genres = dict(genre_frequency.most_common(20))

plt.figure(figsize=(6, 6))
plt.barh(list(top_ten_genres.keys()), list(top_ten_genres.values()), color='#1E0342')
plt.title('Top 20 Most Common Genres on Netflix')
plt.xlabel('Count')
plt.ylabel('Genre')
plt.gca().invert_yaxis()  
plt.show()
```
